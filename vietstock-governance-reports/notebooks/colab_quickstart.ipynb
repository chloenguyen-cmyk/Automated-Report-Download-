## 2. Choose where to save (Google Drive recommended – Colab's own disk is wiped when the session ends)

```python
USE_DRIVE = True   # set False to save in Colab's temporary disk

if USE_DRIVE:
    from google.colab import drive
    drive.mount("/content/drive")
    OUT = "/content/drive/MyDrive/BCTHQT_output"
else:
    OUT = "/content/output"
print("Saving to:", OUT)
```

## 3. Upload your ticker file (.csv or .xlsx)

```python
from google.colab import files

uploaded = files.upload()
TICKERS = list(uploaded)[0]
print("Using:", TICKERS)
```

## 4. Download
Safe to stop and re-run this cell: finished reports are skipped and errors are retried.
Start with a short list (about 10 tickers) to check everything works.

```python
YEARS = "2020-2025"
!govreports download --tickers "{TICKERS}" --out "{OUT}" --years {YEARS}
```

## 5. Which firms have enough years? (optional)

```python
!govreports analyze --out "{OUT}" --min-pairs 3
```

## 6. Pack into ZIP batches (optional)

```python
!govreports zip --out "{OUT}" --eligible-only --tickers-per-batch 25
```
